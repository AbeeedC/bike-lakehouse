# Creating query for gold fact sales table

```python
query = """
SELECT
s.order_number,
c.customer_key,
p.product_key,
s.order_date,
s.ship_date,
s.due_date,
s.sales_amount,
s.quantity,
s.price
FROM workspace.silver.crm_sales s
LEFT JOIN workspace.gold.dim_products p 
    ON s.product_number = p.product_number
LEFT JOIN workspace.gold.dim_customers c
    ON s.customer_id = c.customer_id
"""

df = spark.sql(query)
```

```python
df.limit(10).display()
```

### Writing to Gold table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
