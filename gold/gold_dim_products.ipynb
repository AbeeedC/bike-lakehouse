# Creating query for gold customer dimension table

```python
query = """
SELECT
    ROW_NUMBER() OVER(ORDER BY P.start_date, p.product_number) AS product_key,
    p.product_id,
    p.product_number,
    p.product_name,
    p.product_cost,
    p.product_line,
    p.start_date,
    p.end_date,
    p.category_id,
    pc.category,
    pc.subcategory,
    pc.maintenance_flag
FROM workspace.silver.crm_products p
LEFT JOIN workspace.silver.erp_product_category pc
    ON p.category_id = pc.category_id
--WHERE p.end_date IS NULL --Filters out historical product data and uses live product info
"""

df = spark.sql(query)
```

```python
df.limit(10).display()
```

# Writing to Gold table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```
